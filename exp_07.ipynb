```python
import pandas as pd

df1 = pd.DataFrame({
    'ID':[1,2,3],
    'Name':['A','B','C'],
    'Marks':[80,90,85]
})

df2 = pd.DataFrame({
    'ID':[1,2,3],
    'Grade':['A','A+','A']
})

# 1. Hierarchical Indexing
print(df1.set_index(['ID','Name']))

# 2. Merging
print(pd.merge(df1, df2, on='ID'))

# 3. Combining / Concatenation
print(pd.concat([df1, df2], axis=1))

# 4. Reshaping / Pivoting
print(df1.pivot(index='ID', columns='Name', values='Marks'))
```
